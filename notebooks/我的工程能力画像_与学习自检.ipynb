## 阶段三：工业级 OpenSCENARIO 场景参数化导出 | OpenSCENARIO Exporter

您可以将课堂中调优的典型工况参数一键生成为汽车行业标准的 **ASAM OpenSCENARIO 1.1 (.xosc)** 文件，并下载到本地电脑用于 CARLA / CarMaker / VTD 实车仿真：

```python
from common.xosc_converter import create_cutin_scenario_xosc, validate_xosc_xml, render_xosc_download_button

# 设定您想要导出的加塞工况极限参数
ego_v = 60.0       # 自车巡航速度 (km/h)
target_v = 45.0    # 目标车切入速度 (km/h)
gap_m = 18.0       # 初始切入时距 (m)
cutin_t = 3.0      # 切入时间点 (s)

# 生成并校验工业标准 OpenSCENARIO 格式
xosc_xml = create_cutin_scenario_xosc(
    ego_speed_kmh=ego_v,
    target_speed_kmh=target_v,
    initial_gap_m=gap_m,
    cutin_time_s=cutin_t,
)

is_valid, errs = validate_xosc_xml(xosc_xml)
if is_valid:
    print("✅ OpenSCENARIO 结构化语法与实体校验通过！")
    display(HTML(render_xosc_download_button(xosc_xml, filename="cutin_custom_export.xosc")))
else:
    print("❌ 校验发现异常:", errs)
```
